# Figure 3a

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# load data
dimerization_list = [0, 0.1, 0.2, 0.3]
data_dimerization = dict()
path_to_files = 'Fig3/'
for dimerization in dimerization_list:
    fileName  = path_to_files+'dimerization_'+str(dimerization).replace('.','p') + '.npy'
    data = np.load(fileName, allow_pickle=True).item()
    data_dimerization[dimerization] = data
```

```python
import matplotlib.pyplot as plt

# plot data
color_list = ['#0C5DA5','#00B945','#FF9500','#FF2C00']
Nth_point = 35 # plot every Nth point

fig, ax = plt.subplots(1, dpi =100, sharey=True, figsize=(4, 3))
i = 0
for dimerization, data in data_dimerization.items():

    ax.plot(data['times'], data['boson_observable'],'-', markersize=0.5, color = color_list[i], label = r'$\delta = {}$'.format(dimerization))
    ax.plot(data['times'][::Nth_point], data['spin_observable'][::Nth_point]+14/4,'o', markersize=2, color = color_list[i])
    i+=1
    
ax.set_xticks([0, 0.2,0.4])
ax.set_yticks([0,1,2]) 
ax.set(xlabel=r'$2\pi\,t,\, \mu s$', ylabel=r'$\Delta \mathcal{M}(t)$')

ax.legend()
plt.text(-0.07, 2, r'$(a)$', fontsize=10, color='black')
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_visible(True)
plt.gca().spines['bottom'].set_visible(True)
```

```python
fig.savefig('Figs/figure3a.png', dpi=600, bbox_inches='tight')
```

# Figure 3b

```python
# load data
path_to_file  = 'Fig3/mj.npy'
data = np.load(path_to_file, allow_pickle=True).item()

obs_spin = data['spin_observable']
obs_boson = data['boson_observable']
times = data['times']

L=14 # total number of sites
```

```python
# plot 

Nth_point = 35  # plot every Nth point
color_list = ['#0C5DA5', '#00B945', '#FF9500', '#FF2C00', '#845B97', '#AA3377']
j_list = [0, 5, 6, 7, 8, 13]  # sites to plot


fig, ax = plt.subplots(1, dpi=100, sharey=True, figsize=(4, 3))

solid_lines = []
    
for i, j in enumerate(j_list):
    if j < L//2:
            
        line1 = ax.plot(times, np.array(obs_boson[j]) - 0.5, '-',
                           markersize=0.5, color=color_list[i],
                           label=r'$j = {}$'.format(j+1))[0]
        solid_lines.append(line1)
            
            
        ax.plot(times[::Nth_point], np.array(obs_spin[j][::Nth_point]) - 0.5,
                   'o', color=color_list[i], markersize=2)
    else:
            
        line1 = ax.plot(times, np.array(obs_spin[j]) + 0.5, '-',
                           markersize=0.5, color=color_list[i],
                           label=r'$j = {}$'.format(j+1))[0]
        solid_lines.append(line1)
            
            # Точечный маркер (не будет в легенде)
        ax.plot(times[::Nth_point], np.array(obs_boson[j][::Nth_point]) + 0.5,
                   'o', color=color_list[i], markersize=2)
    
ax.set_xticks([0, 0.2, 0.4])
ax.set_yticks([-0.7, 0, 0.7])
ax.set(xlabel=r'$2\pi\,t,\, \mu s$', ylabel=r'$\Delta m(j,t)$')
    

split_index = 3  
    
# create two legends
legend1 = ax.legend(handles=solid_lines[:split_index], 
                   loc='upper left',
                   bbox_to_anchor=(0, 1.05))  

legend2 = ax.legend(handles=solid_lines[split_index:], 
                   loc='lower left',
                   bbox_to_anchor=(0, -0.05)) 
    

ax.add_artist(legend1)
    
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_visible(True)
plt.gca().spines['bottom'].set_visible(True)

plt.text(-0.09, 0.65, r'$(b)$', fontsize=10, color='black')

plt.show()
```

```python
fig.savefig('Figs/figure3b.png', dpi=600, bbox_inches='tight')
```

# Figure 4

```python
# load data
path_to_file  = 'Fig4/Cj.npy'
data = np.load(path_to_file, allow_pickle=True).item()

obs_spin = data['spin_observable']
obs_boson = data['boson_observable']
times = data['times']
```

# Figure 6

```python
# load data
W_list = [10, 50, 80, 100]

data_W = dict()
path_to_files = 'Fig6/'
for W in W_list:
    fileName = path_to_files + 'W_{}.npy'.format(W)
    data = np.load(fileName, allow_pickle=True).item()
    data_W[W] = data
```

```python
# plot data

Nth_point = 5 # plot every Nth point

color_list = ['#0C5DA5','#00B945','#FF9500','#FF2C00']


fig, ax = plt.subplots(1, dpi =100, sharey=True, figsize=(4, 3))
for i, W in enumerate(W_list): 
        
    ax.plot(data_W[W]['times'], data_W[W]['boson_observable']/16,'-', markersize=0.01, color = color_list[i], label = r'$W = {}$'.format(W))
    ax.plot(data_W[W]['times'][::Nth_point], data_W[W]['spin_observable'][::Nth_point]/16,'o', markersize=2.0, color = color_list[i])
    
ax.set_xticks([0, 3, 6])
ax.set_yticks([0,0.2,0.4]) 
ax.set(xlabel=r'$2\pi\,t,\, \mu s$', ylabel=r'$\mathcal{I}(t)$')

ax.legend(loc=(0.6, 0.55))

plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['left'].set_visible(True)
plt.gca().spines['bottom'].set_visible(True)
plt.show()
```

```python
fig.savefig('Figs/figure6.png', dpi=600, bbox_inches='tight')
```
